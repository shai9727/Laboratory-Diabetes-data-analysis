```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("C:\\Users\\User\\OneDrive\\Desktop\\Diabetes\\Diabetes_data.csv")



```

```python
print(df.head())
```

```python
print(df.info())
```

```python
print(df.describe(include='all'))
```

```python
print(df.isnull().sum())
```

```python
print("duplicates", df.duplicated().sum())
```

```python

df = df.drop_duplicates()
```

```python
df.columns = df.columns.str.strip()
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Remove leading/trailing spaces from column names
df.columns = df.columns.str.strip()

# Standardize categorical variables
df["Gender"] = df["Gender"].astype(str).str.strip().str.title()
df["CLASS"] = df["CLASS"].astype(str).str.strip().str.title()
df["BMI Status"] = df["BMI Status"].astype(str).str.strip().str.title()
# Convert numerical columns
numeric_columns = [
   
    "AGE",
    "Urea",
    "HbA1c",
    "Chol",
    "TG",
    "HDL",
    "LDL",
    "VLDL",
    "BMI",
    
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Check missing values
print(df.isnull().sum())
```

```python
print("\nMean:")
print(df[numeric_columns].mean())

print("\nMedian:")
print(df[numeric_columns].median())

print("\nStandard deviation:")
print(df[numeric_columns].std())
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="CLASS"
)

plt.title("Distribution of Diabetes Classes")
plt.xlabel("Diabetes Class")
plt.ylabel("Number of Patients")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(
    data=df,
    x="Gender",
    hue="CLASS"
)

plt.title("Diabetes Classification by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="CLASS",
    y="HbA1c"
)

plt.title("HbA1c Distribution by Diabetes Class")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="CLASS",
    y="BMI"
)

plt.title("BMI Distribution by Diabetes Class")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,8))

correlation = df[numeric_columns].corr()

sns.heatmap(
    correlation,
    annot= True,
    fmt=".2f"
)

plt.title("Diabetic Parameter Correlation Matrix")
plt.tight_layout()
plt.show()
```

```python
grouped = df.groupby("CLASS")[numeric_columns].mean()

print(grouped)
```
